# Energy Models Workbench

One JupyterLab container with three energy-system models, each in its own
isolated environment:

| Kernel | Model | Upstream |
|---|---|---|
| `pypsa-eur` | PyPSA-Eur — sector-coupled European energy system | [PyPSA/pypsa-eur](https://github.com/PyPSA/pypsa-eur) |
| `eu-flex` | EU-Flex — European flexibility-needs assessment | [EU-ACER/eu-flex-model](https://github.com/EU-ACER/eu-flex-model) |
| `osemosys` | Nour OSeMOSYS Studio | [Islam-Dj/Nour-OSeMOSYS-Studio-Standard](https://github.com/Islam-Dj/Nour-OSeMOSYS-Studio-Standard) |
| `workbench` | This notebook's kernel — no model, just JupyterLab tooling | — |

Switch kernels from the kernel selector in the top-right corner. Each kernel
runs in its own environment, so a dependency clash in one model cannot break
another.


## What is mounted where


```python
from pathlib import Path

for path in [
    "/work/models",
    "/work/examples",
    "/work/osemosys",
    "/data",
    "/results",
]:
    p = Path(path)
    print(f"{path:20s} exists={p.exists()}")

print("\nModel checkouts:")
for repo in sorted(Path("/work/models").iterdir()):
    if repo.is_dir():
        head = (repo / ".hem-ref")
        sha = head.read_text().strip()[:12] if head.exists() else "?"
        print(f"  {repo.name:20s} {sha}")
```

## Run the models

Each model has a short example notebook in `/work/examples`. The model
repositories themselves live in `/work/models` and are writable, so you can
create configs and run workflows there without recreating the container.

The run scripts are on `PATH`:

- `run-pypsa-eur.sh` — plan or run the PyPSA-Eur test workflow.
- `run-eu-flex.sh` — run the EU-Flex teaser workflow.
- `start-osemosys-studio.sh` — start the Studio server (already started by the
  container entrypoint on port 8775, proxied at `/proxy/8775/`).

Data downloads go to `/data`; solve outputs go to `/results` or into the
model's own `runs/` folder. Both survive container rebuilds.
